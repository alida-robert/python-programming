# string operations

```python
#slicing
s="python"
print(s[0])  #index 0
print(s[3:6])   #from index 3 to index 5
print(s[1:5])
print(s[-6:])
print(s[:4])
print(s[-5:-1]) 
print(s[1:7])
print(s[0:])
a="1234567"
print(a[0:8:2])
print(a[1:8:2])
print(a[0:8:1])
print(a[::-1])   #reverse
```

## string methods

```python
a="hellO worLD"
print(a.upper())
print(a.lower())
print(a.capitalize())
print(a.title())
print(a.swapcase())
b="straße"
print(b.casefold())
c="apple"
print(c.find("p"))
print(c.find("b"))    # if not found shows -1 
print(c.index("l"))
#print(c.index("b"))  # value error
d="banana"
print(d.rfind("n"))   
print(d.count("a"))



```

```python
text="hello world"
print(text.startswith("hello"))
print(text.endswith("hello"))
print(text.isalpha())
a="abcd"
print(a.isalpha())
print(a.isdigit())
b="1234"
print(b.isalpha())
print(b.isdigit())






```

```python
x="hello"
print(x.islower())
print(x.isupper())
y="HELLO"
print(y.isupper())
print(y.islower())
```

```python
text="python  is  fun"
print(text.split())
print(text.split("n"))
word=text.split()
print('_'.join(word))
x="hello_world"
print(x.partition('l'))
print(x.partition('_'))
print(x.partition('o'))
a=" hello world   "
print(a.strip())
print(a.rstrip())
print(a.lstrip())

h="67"
print(h.zfill(5))

print('56'.zfill(9))



```

```python
a="hello world"
print(a.replace("hello","hiii"))
b="I love python"
print(b.replace("love","like"))
print("apple\norange")
print("apple\torange")
print("apple\\orange")
```

### String formatting

```python
name="alice"
age=20
print("name:%s,age:%d"%(name,age))
print('name:{},age:{}'.format(name,age))
print(f'name:{name},age:{age}')
pi=3.1454567
print(f'pi with 3 decimals:{pi:3f}')
```
