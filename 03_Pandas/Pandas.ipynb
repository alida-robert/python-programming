## 1. What is Pandas?
Pandas is a Python library for working with tabular data. Its two main structures are **Series** (one-dimensional) and **DataFrame** (rows and columns).

Install if needed: `pip install pandas`

```python
import pandas as pd
print("Pandas version:", pd.__version__)
```

## 2. Series: creating and accessing
A Series is a labeled, one-dimensional collection.

```python
s = pd.Series([10, 20, 30])
print(s)
```

```python
print("By position:", s.iloc[1])
```

## 3. DataFrame: creating data
A DataFrame is a two-dimensional table. We will reuse this sample student dataset.

```python
df = pd.DataFrame({"Name": ["Anu", "Binu", "Charu", "Deepa", "Ebin", "Fathima"], "Age": [20, 21, 19, 22, 20, 21], "Department": ["BCA", "BSc", "BCA", "BSc", "BCA", "BSc"], "Marks": [85, 72, 90, 65, 72, 95], "Attendance": [90, 80, 95, 70, 85, 98]})
display(df)
```

## 4. Inspecting data
Use head, tail, shape, columns, dtypes, info and describe to understand a dataset.

```python
display(df.head(3))
display(df.tail(2))
print("Shape:", df.shape)
print("Columns:", df.columns)
df.info()
display(df.describe())
```

## 5. Selecting columns
Select one column with brackets or multiple columns with a list.

```python
print(df["Name"])
display(df[["Name", "Marks"]])
```

## 6. Selecting rows: loc and iloc
`loc` selects by label; `iloc` selects by integer position.

```python
print(df.loc[0, "Name"])
display(df.loc[1:3, ["Name", "Marks"]])
display(df.iloc[0:3, 1:3])
```

## 7. Filtering rows
Use comparison operators and `&` / `|` with parentheses.

```python
display(df[df["Marks"] >= 80])
display(df[(df["Department"] == "BCA") & (df["Attendance"] > 85)])
```

## 8. Adding and updating columns
Pandas performs operations across whole columns without loops.

```python
df["Bonus"] = 5
df["Final_Marks"] = df["Marks"] + df["Bonus"]
def grade(marks):
    if marks >= 95:
        return "O"
    elif marks >= 90:
        return "A"
    elif marks >= 85:
        return "B"
    elif marks >= 80:
        return "C"
    elif marks >= 75:
        return "D"
    else:
        return "F"
df["Grade"] = df["Final_Marks"].apply(grade)
df["Result"] = df["Final_Marks"].apply(lambda x: "Pass" if x >= 75 else "Fail")
display(df)
```

## 9. Sorting
Sort values ascending or descending.

```python
display(df.sort_values("Marks"))
display(df.sort_values("Marks", ascending=False))
```

## 10. Statistics and aggregations
Common functions: sum, mean, median, min, max, std, count, value_counts, nunique.

```python
print("Total:", df["Marks"].sum())
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Min / Max:", df["Marks"].min(), df["Marks"].max())
print("Standard deviation:", df["Marks"].std())
print("Count:", df["Marks"].count())
print(df["Department"].value_counts())
```

## 11. GroupBy operations
Group rows by category and summarize each group.

```python
display(df.groupby("Department")["Marks"].mean())
display(df.groupby("Department").agg(Students=("Name", "count"), Average_Marks=("Marks", "mean"), Highest_Marks=("Marks", "max")))
```

## 12. Missing values
`isna`, `fillna`, and `dropna` help handle missing data.

```python
Norm = pd.DataFrame({"Name": ["Anu", "Binu", "Charu", "Deepa"], "Marks": [85, None, 90, None], "City": ["Kozhikode", None, "Kochi", "Kochi"]})
display(Norm)
print("Missing per column:", Norm.isna().sum())
filled = Norm.copy()
filled["Marks"] = filled["Marks"].fillna(filled["Marks"].mean())
filled["City"] = filled["City"].fillna("Unknown")
display(filled)
display(Norm.dropna())
```

## 13. Duplicates
Identify and remove duplicate records.

```python
dupes = pd.DataFrame({"Name": ["Anu", "Binu", "Anu"], "Marks": [85, 72, 85]})
print("Duplicate rows:", dupes.duplicated().sum())
display(dupes.drop_duplicates())
```

## 14. Renaming and dropping
Rename columns and remove unwanted rows/columns.

```python
renamed = df.rename(columns={"Marks": "Exam_Marks"})
display(renamed.head(4))
display(df.drop(columns=["Bonus"]).head(4))
display(df.drop(index=[0]).head(4))
```

## 16. Replacing values and mapping
Use replace for substitutions and map for category mappings.

```python
print(df["Department"].replace({"BCA": "Computer Applications"}))
df["Dept_Code"] = df["Department"].map({"BCA": 1, "BSc": 2})
display(df[["Name", "Department", "Dept_Code"]])
```

## 17. Merging tables
`merge` joins two DataFrames using a shared key.

```python
fees = pd.DataFrame({"Name": ["Anu", "Binu", "Charu"], "Fees_Paid": [True, False, True]})
display(df.merge(fees, on="Name", how="left"))
```

## 18. Concatenating tables
`concat` stacks tables vertically or horizontally.

```python
part1 = df.iloc[:3][["Name", "Marks"]]
part2 = df.iloc[3:][["Name", "Marks"]]
display(part1)
display(part2)
display(pd.concat([part1, part2]))
```

## 19. Pivot tables and crosstabs
Pivot tables summarize values by categories; crosstab counts combinations.

```python
display(pd.pivot_table(df, values="Marks", index="Grade", aggfunc="sum"))
display(pd.crosstab(df["Department"], df["Result"]))
```

## 20. Reading and writing CSV
Create a CSV, read it back, and export processed data. These files are written to the notebook working directory.

```python
df.to_csv("students_pandas_example.csv", index=False)
loaded = pd.read_csv("students_pandas_example.csv")
display(loaded.head())
```

## 21. Working with dates
Convert strings to datetime, then extract year, month, and day.

```python
dates = pd.DataFrame({"Date": ["2026-01-10", "2026-02-15", "2026-03-20"], "Sales": [1000, 1200, 900]})
dates["Date"] = pd.to_datetime(dates["Date"])
dates["Month"] = dates["Date"].dt.month_name()
dates["Day_Name"] = dates["Date"].dt.day_name()
dates["Date"] = dates["Date"].dt.strftime("%d/%m/%Y")
display(dates)
```

Pandas can plot charts with Matplotlib installed.

```python
import matplotlib.pyplot as plt
df.groupby("Department")["Marks"].mean().plot(kind="bar", title="Average marks by department")
plt.ylabel("Average marks")
plt.show()
```
