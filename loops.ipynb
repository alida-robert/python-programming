# loop through collections

```python
number=[1,2,3,4,5]
for i in number:
    print(i)
```

```python
num=(1,2,6,10,8)
for j in num:
    print(j)
```

```python
n={1,4,9,3,4,}
for i in n:
    print(i)
```

```python
d={1:"a",2:"b"}
for i,h in d.items():
    print(i,h)
for j in d.values():
    print(j)
for k in d:
    print(k)
for l in d.items():
    print(l)
```

```python
li=[]
for i in range(1,11):
    li.append(i)
print(li)
```

```python
se=set()
for i in range(1,11,2):
    se.add(i)
print(se)
```

```python
parts=[]
for i in range(1,9):
    parts.append('x')
print(parts)
    
```

```python
for i in range(1,4):
    for j in range(3,6):
        print(i,j)
```

```python
for i in range(5):       #row
    for j in range(8):     #column
        print(i,j,end=" ")
    print()
```

```python
for i in range(5):               
    for j in range(i):
        print('*',end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print(j,end=" ")
    print()
```

```python
# loop with else
for i in range(3):
    print(i)
else:
    print("complete")
```

```python
count=0
for i in range(2,21,2):
    print(i)
    count+=1
print("count:",count)
```

```python
count=0
for i in range(1,21):
    if i%2==0:
        print(i)
        count+=1
print("count:",count)   
```

```python
sum=0
for i in range(1,11):
    print(i)
    sum+=i
print("sum:",sum)
```

# While loop

```python
count=5
while count>0:
    print(count)
    count-=1
print("Liftoff")
```

```python
count=1
while count<=10:
    print(count)
    count+=1
```

```python
count=10
while count>=1:
    print(count)
    count-=1
```

```python
i=1
sum=0
while i<=10:
    sum=sum+i
    i+=1
print(sum)
```

```python
i=1
while i<=100:
    if i%5==0:
        print(i)
    i+=1
```

```python
n=546
while n>0:
    num=n%10
    n=n//10
    print(num,end=" ")
```

```python
l=[1,2,3,4,5,6,7]
even=0
odd=0
i=0
while i<len(l):
    if l[i]%2==0:
        even+=1
    else:
        odd+=1
    i+=1
print(even)
print(odd)
    
    
    
```

```python
i=1
while i<=5:
    print("*"*i)
    i+=1
```

```python
i=5
while i>0:
    print("*"*i)
    i-=1
```

```python
i=1
while i<=5:
    print(" "*(5-i)+"*"*(2*i-1))
    i+=1
```

```python
i=10
while i>=0:
    print(" "*(10-i)+"*"*(2*i-1))
    i-=1
    
```

```python
n=5
i=1
while i<=10:
    print(f"{i}*{n}={i*n}")
    i+=1
```

# loop control statements

```python
for i in range(10):
    if i==3:
        break
    print(i)
```

```python
for n in range(6):
    if n ==3:
        continue
    print(n)
```

```python
for n in range(6):
    if n==2:
        pass
    print(n)
```

```python
num=[2,4,6,10,99]
for i in num:
    if i>6:
        print(f"stopped at {i}")
        break
```

```python
li=[2,45,88,1,10]
value=88
for i in li:
    if i==value:
        print("found")
        break
        
else:
    print("not found")
```

```python
for i in range(1,11):
    if i%2!=0:
        continue
    print(f"{i}-even")
```

```python
name=["anna","alice","bob","john"]
for i in name:
    if i=="bob":
        continue
    print(f"Hi {i}")
```

```python
password=2005
attempt=3
while attempt>0:
    user_pass=int(input("enter password"))
    if user_pass==password:
        print("login")
        break
    else:
        print("Invalid password please try again")         
    attempt-=1
else:
    print("account blocked")
```

```python
booked=[1,3,5,10,20,15]                     
for i in range(1,21):
    if i in booked:
        continue
    else:
        print("available seats:",i)
```
