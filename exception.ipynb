```python
#multiple exception
def process(value):
    try:
        number=int(value)
        result=100/number
        print("result:",result)
    except ValueError:
        print(f"{value} is not a valid number")
    except ZeroDivisionError:
        print("cannot divide by zero")
process(10)
process(0)
process("abc")
```

```python
#else
def check_positive(num):
    try:
        assert num>0
    except AssertionError:
        print(num,"not postive")
    else:
        print(num,"positive")
check_positive(-1)
```

```python
def div(a,b):
    try:
        result=a/b
    except ZeroDivisionError:
        print("cannot divide by zero")
    else:
        print(result)
div(10,2)
div(10,0)
```

```python
#finally
def divide(a,b):
    try:
        result=a/b
    except ZeroDivisionError:
        print("cannot divide by 0")
    else:
        print("division successful",result)
    finally:
        print("execution complete")
divide(10,2)
print()
divide(10,0)
```
