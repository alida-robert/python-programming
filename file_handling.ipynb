# file handling

```python
#read and write
with open("demo.txt","w") as f:
    f.write("first write")
with open("demo.txt","r") as f:
    print(f.read())
```

```python
#append
with open("a_demo.txt","a") as f:
    f.write("Hello \n")
    f.write("welcome")
with open("a_demo.txt","r") as f:
     print(f.read())
```

```python
#append multiple times using loop
for i in range(1,4):
    with open("abc.txt","a") as f :
        f.write(f"entry{i}\n")
with open("abc.txt","r") as f:
    print(f.read())
```

```python
for i in range(1,4):
    with open("ab.txt","w") as f :
        f.write(f"entry{i}\n")
with open("ab.txt","r") as f:
    print(f.read())
```

```python
#x mode for already existing file
import os
if os.path.exists("a_demo.txt"):
    os.remove("a_demo.txt")
with open("a_demo.txt","x") as f:
    f.write("created using x mode")
print("file created successfully with x mode")
with open("a_demo.txt") as f:
    print(f.read())


```

```python
#for existing file x mode raise error
with open("a_demo.txt","x") as f:
    f.write("created using x mode")
```

```python
#create a file using x mode

with open("a1_demo.txt","x") as f:
    f.write("created using x mode")
with open("a1_demo.txt") as f:
    print(f.read())
```

```python
try:
    with open("abcde_demo.txt", "x") as f:
        f.write("created using x mode")
except FileExistsError as e:
    print(e)
    print("The file already exists, so x mode failed!")
```

```python
#a+
with open("a+_demo.txt","w")as f:
    f.write("existing content \n")
with open("a+_demo.txt","a+") as f:
    f.write("appended content")
    f.seek(0)
    print(f.read())
```

```python
#reading and writing text files using with open()
lines=["apple\n","orange\n","cherry"]
print(lines)
with open("fruit.txt","w")as f:
    f.writelines(lines)
with open("fruit.txt","a") as f:
    f.writelines(lines)
print("fruit file created successfully")
with open("fruit.txt") as f:
    print(f.read())          #read()

```

```python
#remove
with open("fruit.txt","r") as f:
    lines=f.readlines()
del lines[2]
with open("fruit.txt","w") as f:
    f.writelines(lines)
with open("fruit.txt","r") as f:
    print(f.read())
```

```python
#cleanup
import os
for i in ["abcd_demo.txt","a1_demo.txt"]:
    if os.path.exists(i):
        os.remove(i)
print("files cleaned up")
```
