# operators

### Arithmetic operators

```python
a=10
b=4
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)

```

### comparison operator

```python
a=10
b=5
c=10
print(a==b)
print(a==c)
print(a>b)
print(a>b<c)
print(b>=c)
print(c<=b)
```

### assignment operator

```python
n=10
n+=1   #n=n+1
print(n)
n-=3
print(n)
n*=2
print(n)
n/=2
print(n)
n%=3
print(n)
# n**=2
n=n**2
print(n)
```

## identity operators

```python
A=100
B=A
print(A is B)
print(A is not B)
print(id(A))
print(id(B))

C=300
D=300
print(C is D)
print(C is not D)
print(id(C))
print(id(D))

a1=[1,3,5]
b1=a1         # same object
c1=[1,3,5]
print(a1 is b1)
print(a1 is c1)
```

## membership operator

```python
text="apple"
print("l" in text)
print("o" in text)
print("o" not in text)
print("e" not in text)
```

## logical operator

```python
a=3
b=10
print(a<b and b>a)          # and
print(a>b and b>a)                          
print(a>b or b>a)           #or
print(not(a>b and b>a))     # not
```

## bitwise operator

```python
a=10
b=4
print(a&b)  # and
print(a|b)   # or
print(a^b)  #XOR
print(~a)   # not
print(a<<2) #leftshift
print(a>>2)  #right shift
```
