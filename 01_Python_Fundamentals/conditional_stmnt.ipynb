## if elif else

```python
num1=int(input("enter the number:"))
num2=int(input("enter number"))
if num1==num2:
    print("equal")
elif num1>num2:
    print("num1 is greater")
else:
    print("num2 is greater")
```

```python
a=int(input("enter a:"))
b=int(input("enter b:"))
c=int(input("enter c:"))
if a>b and a>c:
    print("a is largest")
elif b>a and b>c:
    print("b is largest")
else:
    print("c is largest")
```

```python
signal=input("enter signal:").lower()
if signal=="red":
    print("stop")
elif signal=="yellow":
    print("slow")
elif signal=="green":
    print("go")
else:
    print("invalid")
```

## nested if

```python
score=int(input("enter score"))
if score>=30:
    if score>=90:
        print("student passed with distinction")
    else:
        print("student passed")
else:
    print("fail")
```

```python
num=int(input("enter number"))
if num>0:
    if num%2==0:
        print(f"{num} is even and positive number")
    else:
        print(f"{num} is odd and positive number")
else:
    print("number is negative")
```

```python
age=int(input("enter age"))
income=int(input("enter income"))
if age>18:
    if income<=10000:
        print("eligible for scholarship")
    else:
        print("not eligible")
else:
    print("not eligible for scholarship")


```
