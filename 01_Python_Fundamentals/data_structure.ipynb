# Dictionary

```python
d={"name":"Alice","age":28}
d_1=dict(a=1,b=2,c=3)  
print(d_1)
print(d)
print(d["name"])     #access
print(d_1["a"])
di={} #creates empty dictionary

#key cannot be same it print last  repeated key 
d_3={"name":"Alice","age":28,"place":"chennai","name":"bob"}
print(d_3)
d_3["place"]="mumbai"   #change value

d_3["course"]="bca"
print(d_3)
print(d_3.get("age"))
print(d_3["name"])

#pop
d_3.pop("age")
print(d_3)
#popitem
d_3.popitem()
print(d_3)  
del d_3["place"]
print(d_3)
del d

d_3.clear()
print(d_3)
```

## nested dictionary

```python
student={
    "s1":{"name":"alice","age":22},
    "s2":{"name":"bob","age":28,"city":"mumbai"}
}
print(student["s1"]["name"])
print(student["s2"]["city"])
print(student["s1"]["name"],student["s1"]["age"])
print(student["s1"])
```

```python
c={
    "child1":{"name":"anna",
              "age":15,
              "hobby":"singing"
    },

    "child2":{"name":"abi",
              "age":18,
              "hobby":"dancing"
    },
    "child3":{"name":"aleena",
              "age":14,
              "hobby":"reading"
    }
}
print(c)
print(c["child1"])
print(c["child2"]["hobby"])
print(c["child3"]["name"])
print(c["child1"]["name"],c["child1"]["age"],c["child1"]["hobby"])
```

## dictionary comprehension

```python
# create dictionary from list
numbers=[1,2,3,4]
square={x:x*x for x in numbers}
cube={x:x*x*x for x in numbers}
s={x:x+10 for x in numbers}
print(square)
print(cube)
print(s)
```
