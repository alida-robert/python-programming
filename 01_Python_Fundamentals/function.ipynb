```python
#variable length arguments(*args)
def add(*numbers):
    total=0
    for i in numbers:
        total+=i
    print(f"sum : {total}")
add(2,3)
add(1,2,3,4,5)
```

```python
def show_item(*items):
    for i in items:
        print(i)
show_item("Apple","Orange","kiwi","mango",89,23)
```

```python
#combine fixed and variable argument
def greet(msg,*names):
    for n in names:
        print(msg,n)
greet("Hello","Anna","Mariya","liya")
```

```python
# keyword argument(**kwargs)
def info(**details):
    for key,value in details.items():
        print(key,":",value)
info(name="Aswathi",age=27,place="clt",course="Python")
```

```python
# fixed and keyword argment
def show_detail(id,**info):
    print("ID:",id)
    for key,values in info.items():
        print(key,":",values)
show_detail(101,name="abi",place="tvm",age=30)
```

```python
#unknown number of arguments
def mix_arg(*args,**kwargs):
    print("Positional:",args)
    print("Keyword:",kwargs)
mix_arg(1,2,3,name="Alida",subject="DS")
```

# Local and global variable

```python
# local variable
def test():
    x=100
    print(x)
test()
```

```python
# global variable
x=200
def test():
    print(x)
test()
print(x)
```

```python
def loc():
    x="local"
    print("Inside variable:",x)
loc()
```

```python
#enclosing scope-a variable inside an outer fn that is accessible to an inner fn
def outer():
    x="enclosing"               
    def inner():
        print("Inside inner",x) # x -enclosing coz it is inside the fn surrounding inner()
    inner()
outer()
```

```python
#global
y="global"
def glo():
    print("Inside fn:",y)
glo()
print("outside fn:",y)
```

```python
#LEGB rule
a="global"
def outer():
    a="enclosing"
    
    def inner():
        a="local"
        print("inner:",a)
    inner()

    print("outer:",a)
outer()

print("global:",a)
```

## special keyword
global-modify global variable
nonlocal-modify enclosing variable

```python
b="global"
def outer():
    b="enclosing"
    def inner():
        nonlocal b
        b="modified enclosing"
    inner()
    print("outer:",b)
outer()
print("global:",b)
```

```python
#global keyword
c="global"
def outer():
    c="enclosing"
    def inner():
        global c
        c="modified enclosing"
    inner()
    print("outer:",c)
outer()
print("global:",c)
```

## Recursive

```python
def factorial(n):
    if n==1:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))
```

```python
def power(a,b):
    if b==0:
        return 1
    else:
        return 2*power(a,b-1)
print(power(2,3))
```

```python
def countdown(n):
    if n==0:
        print("done")
    else:
        print(n)
        countdown(n-1)
(countdown(5))
```

```python
def countdwn(n):
    if n==0:
        return 0
    else:
        print(n)
        return countdwn(n-1)
print(countdwn(5))
```

## lambda function
anonymous fn

```python
#lambda argument:expression

square=lambda x:x*x
print(square(9))
```

```python
add=lambda a,b:a+b
print(add(5,4))
```

```python
multi=lambda a,b:a*b
print(multi(3,2))
```

```python
oddeven=lambda a:"even" if a%2==0 else "odd"
print(oddeven(4))
print(oddeven(9))
```

## map()

```python
num=[1,2,3,4]
square=list(map(lambda x:x*2,num))
print(square)
```

## filter()

```python
numbers=[1,2,3,4,5,6]
evens=list(filter(lambda x:x%2==0,numbers))
print(evens)
```

## sorted()

```python
nums=[1,4,3,2]
square=list(sorted(map(lambda x:x*x,nums)))
print(square)
```
